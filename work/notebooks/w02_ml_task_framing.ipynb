## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# The target we are predicting is `is_declining_label`.
# Crucial Gotcha: Because this label is derived from `trend_direction` (which itself comes from `trend_pct`), we must strictly exclude `trend_direction` and `trend_pct` from our features to avoid the label trap.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

# Loading the lane's slice of the starter data
url = 'https://raw.githubusercontent.com/muhammadabdurrehmanmaqsood/flyrank-ml-internship-abdurrehman/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

# 4.1) DEBUG: Let's see exactly what columns are in your dataset
print("Available columns:", df.columns.tolist())

# 4.2) Derive the target label BEFORE dropping the trend columns
# (You may need to adjust the string 'Declining' based on what unique values are actually in trend_direction)
if 'trend_direction' in df.columns:
    print("\nUnique values in trend_direction:", df['trend_direction'].unique())
    # Assuming the string is 'Declining' or 'declining'
    df['is_declining_label'] = (df['trend_direction'] == 'Declining').astype(int)

# 4.3) The unit of analysis: One row = one pseudonymized content item.
# Now we drop the derived trend columns to strictly avoid the label trap.
safe_df = df.drop(columns=['trend_direction', 'trend_pct'], errors='ignore')

print("\nUnit of Analysis (1 row = 1 piece of content):")
display(safe_df.head(3))

# 4.4) Sketching the target column
if 'is_declining_label' in safe_df.columns:
    print("\nTarget Column Sketch (is_declining_label distribution):")
    display(safe_df['is_declining_label'].value_counts(dropna=False))
else:
    print("\nTarget label could not be created. Please check the 'Available columns' list above to find the correct target name.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# A simple fixed rule (like "flag the page if traffic drops by 10%") isn't
# enough because content performance data is extremely messy. Signals like `ctr`
# and `engagement_rate` behave differently across different content types, and
# some metrics like `scroll_rate` can legitimately exceed 100 due to mismatched
# measurement systems. ML earns its place here because it can weigh these
# complex, tangled, and shifting signals far better than a handwritten
# if-statement.
```
