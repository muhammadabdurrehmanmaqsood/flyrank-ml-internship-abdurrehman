*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

```python
# Selected Lane: Lane 2: Refresh / Content Opportunity Scoring

# Why: The goal is to build a decision-support tool that prioritizes our content team's backlog.
# Instead of relying on guesswork to figure out which pages are losing traffic, we can use
# observable search signals to identify pages that are currently declining but still have
# enough baseline demand to be worth fixing. This translates a raw data problem into an
# actionable, ranked queue for human reviewers.
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# The Decision: Which specific, currently declining pages should the content team review first today?

# The Action: A human reviewer opens the top-ranked page in the queue
# and decides whether to update the content, consolidate it with another page,
# or leave it alone based on the provided reason codes.

# The Cost of a Wrong Call: If the model ranks a low-value, strictly seasonal,
# or permanently dead page at the top of the queue, a reviewer wastes valuable hours
# working on a page that has no potential to drive meaningful traffic,
# ignoring higher-value opportunities.
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the dataset directly from the raw GitHub URL
url = 'https://raw.githubusercontent.com/muhammadabdurrehmanmaqsood/flyrank-ml-internship-abdurrehman/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

# 1. Total pages in the dataset
total_pages = len(df)

# 2. How many pages are currently flagged as declining?
declining_pages = len(df[df['trend_direction'] == 'down'])

# 3. Of those declining pages, how many have meaningful traffic at risk? (e.g., > 500 impressions)
high_value_declining = len(df[(df['trend_direction'] == 'down') & (df['impressions_90d'] > 500)])

print(f"Total pages in starter dataset: {total_pages}")
print(f"Pages showing a downward trend: {declining_pages}")
print(f"Declining pages with >500 recent impressions (our primary targets): {high_value_declining}")
```

To justify this lane, I checked the starter dataset (data/raw/content_refresh_anonymized.csv):



1.   There are **30000** total pages in the starter slice.
2.   Of those, **16262** pages are actively showing a downward trend.
3.   More importantly, **9956** of those declining pages have generated over **500** impressions in the last **90** days. This proves there is a substantial, high-value backlog that requires prioritization, rather than just a handful of dead pages.
